```python
import matplotlib.pyplot as plt  # 导入绘图库

# 1. 准备数据（x轴和y轴的数据，通常用列表或numpy数组）
x = [1, 2, 3, 4, 5, 6]  # x轴数据（比如时间、序号）
y = [2, 4, 6, 5, 7, 9]  # y轴数据（对应x的数值）

# 2. 绘制折线图
plt.plot(x, y)  # 核心函数：用x和y绘制折线

# 3. 添加图表信息（可选但推荐）
plt.xlabel("x axis label")  # x轴名称
plt.ylabel("y axis label")  # y轴名称
plt.title("Basic Line Graph Sample")  # 图表标题

# 4. 显示图像
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [2, 4, 6, 5, 7, 9]

# 自定义样式的折线
plt.plot(
    x, y,
    color='green',       # 线条颜色：绿色
    linestyle='-',      # 线型：虚线
    marker='o',          # 数据点标记：圆点
    linewidth=2,         # 线条粗细
    markersize=8,        # 标记点大小
    markerfacecolor='yellow'  # 标记点填充色
)

plt.xlabel("x axis label")
plt.ylabel("y axis label")
plt.title("Basic Line Graph Sample")
plt.show()
```
