```python
import matplotlib.pyplot as plt  # 导入绘图库

# 1. 准备数据（x轴和y轴的数据，通常用列表或numpy数组）
x = [1, 2, 3, 4, 5, 6]  # x轴数据（比如时间、序号）
y = [2, 4, 6, 5, 7, 9]  # y轴数据（对应x的数值）

# 2. 绘制折线图
plt.plot(x, y)  # 核心函数：用x和y绘制折线

# 3. 添加图表信息（可选但推荐）
plt.xlabel("x axis label")  # x轴名称
plt.ylabel("y axis label")  # y轴名称
plt.title("Basic Line Graph Sample")  # 图表标题

# 4. 显示图像
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [2, 4, 6, 5, 7, 9]

# 自定义样式的折线
plt.plot(
    x, y,
    color='green',       # 线条颜色：绿色
    linestyle='--',      # 线型：虚线
    marker='o',          # 数据点标记：圆点
    linewidth=2,         # 线条粗细
    markersize=8,        # 标记点大小
    markerfacecolor='yellow'  # 标记点填充色
)

plt.xlabel("x axis label")
plt.ylabel("y axis label")
plt.title("Basic Line Graph Sample")
plt.show()
```

```python
import matplotlib.pyplot as plt

# 数据：两组y值，共享同一个x轴
x = [1, 2, 3, 4, 5, 6]
y1 = [2, 4, 6, 5, 7, 9]  # 第一组数据
y2 = [3, 5, 4, 6, 8, 7]  # 第二组数据

# 绘制第一条线（标记为“数据1”）
plt.plot(x, y1, color='blue', marker='o', label='data 1')
# 绘制第二条线（标记为“数据2”）
plt.plot(x, y2, color='red', marker='s', label='data 2')

# 添加图例（显示label的内容）
plt.legend()

# 其他信息
plt.xlabel("time")
plt.ylabel("value")
plt.title("Multiple Datasets")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [2, 4, 6, 5, 7, 9]

plt.plot(x, y, color='purple')

# 固定x轴范围：0到7
plt.xlim(0, 7)
# 固定y轴范围：0到10
plt.ylim(0, 10)

plt.xlabel("x axie")
plt.ylabel("y axis")
plt.title("Fixed Axis Range")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [2, 4, 6, 5, 7, 9]

plt.plot(x, y, color='orange', marker='^')

# 添加网格线（linestyle设置线型，alpha设置透明度）
plt.grid(linestyle='--', alpha=0.7)

plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("Line Graph With Grid")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

# 绘制折线
plt.plot(x, y, color="black", linewidth=1.5)

# 1. 填充x轴上方区域（y >= 0）
plt.fill_between(
    x, y, 0,
    where=(y >= 0),  # 只填充y>=0的部分
    color="red",
    alpha=0.3,
    label="Upper Area"
)

# 2. 填充x轴下方区域（y < 0）
plt.fill_between(
    x, y, 0,
    where=(y < 0),  # 只填充y<0的部分
    color="blue",
    alpha=0.3,
    label="Below Area"
)

# 其他设置
plt.xlabel("x")
plt.ylabel("y = sin(x)")
plt.title("Fill Different Colors Between Either Lines And X Axis")
plt.legend()  # 显示标签
plt.grid(linestyle="--", alpha=0.3)
plt.show()
```
